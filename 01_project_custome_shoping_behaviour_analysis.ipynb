```python
data["Review Rating"]=data.groupby("Category")["Review Rating"].transform(lambda x:x.fillna(x.median())) #replacig the
                                                                                                          #missing review by median of the review of this category.
data.isnull().sum()
```

```python
#all columns name should be lowercase and underscore for easy use. ie:snakecasing ho re
data.columns=data.columns.str.lower()
data.columns=data.columns.str.replace(" ","_")

print(data)
```

```python
data=data.rename(columns={"purchase_amount_(usd)":"purchase_amount"})
data.columns
```

```python
#create a column age_group
labels=["Young Adult","Adult","Middle age","senior"]
data["age_group"]=pd.qcut(data["age"],q=4,labels=labels)  #qcut function divide the age into 4 quaters and assign levels as we provide above
```

```python
data[["age","age_group"]].head(15)
```

```python
#create column purchase_frequency_days
frequency_mapping={
"Fortnightly":14,
"Weekly":7,
"Monthly":30,
"Quarterly":90,
"Bi-Weekly":14,
"Annually":365,
"Every 3 Months":90
}
data["purchase_frequency_days"] = data["frequency_of_purchases"].map(frequency_mapping)
```

```python
from sqlalchemy import create_engine
import pandas as pd

# ── 1. Load your data first ──────────────────────────────
data = pd.read_csv("customer_shopping_behavior.csv")
print("Before Renaming:\n", data.columns.tolist())

# ── 2. Rename columns to lowercase + snake_case ──────────
data.columns = data.columns.str.lower()
data.columns = data.columns.str.replace(" ", "_")
data.columns = data.columns.str.replace("(", "").str.replace(")", "")
print("\nAfter Renaming:\n", data.columns.tolist())

# ── 3. Connect to PostgreSQL ─────────────────────────────
host = "localhost"
port = 5432
database = "customer_behavior"
username = "postgres"
password = "badal12"

engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}")

# ── 4. Load renamed DataFrame into PostgreSQL ────────────
table_name = "customer"
data.to_sql(table_name, engine, if_exists="replace", index=False)
print(f"\nData loaded successfully into table '{table_name}' in database '{database}'.")

# ── 5. Read back from PostgreSQL to verify ───────────────
df = pd.read_sql(f"SELECT * FROM {table_name};", engine)
print(df.head())
```
